## Pandas Introduction

```python
import pandas as pd
```

```python
# An example of creating a Dataframe using a dictionary
info = {
    "Name:": ["John", "David", "Thomas"],
    "CGPA:": [8.9, 9.0, 9.2]
}

dataf = pd.DataFrame(info)
print(dataf)
```

```python
dataf
```

```python
# Creating a series using a list
# Index and Labels are same by default
s = pd.Series([1, 2, 3, 4, 5])
print(s)
print(type(s))

# Indexing
print(s[0])
print(s[2])
print(s[0:3]) # using slicing

# Creating custom labels
print("---------------------------------")
s1 = pd.Series([22, 25, 27, 29, 24], index=["John", "David", "Thomas", "Ross", "Kramer"])
print(s1)

# Indexing
print(s1["John"])
print(s1["Thomas"])
#print(s1[0]) # Can do this too, just get a warning not an error
#Generally we don't use custom labels and indexing together
print(s1.index)
```

```python
# Series properties
s2 = pd.Series([1, 2, 3, 4, 5])
print(s2)

new_s2 = s2.drop(1)
print(s2)
print(new_s2)

#Other properties are same as in numPy arrays:
```

```python
# DataFrame in pandas
data = {
    "Name:": ["Thomas", "John", "Henry"],
    "Age:" : [24, 25, 27],
    "CGPA:" : [9.1, 8.9, 9.0]
}

df = pd.DataFrame(data)
print(df)
print(type(df))
print("-------------")
print(df.index)
print(df.columns)
```

```python
df
```

```python
# Creating dataFrame using list of lists
data1 = pd.DataFrame([["Adam", 22], ["Jack", 23], ["Tim", 25]], columns=["Name", "Age"])
print(data1)
```

```python
# Creating using numPy arrays
import numpy as np
num_arr = np.array([[1, 2, 3], [4, 5, 6]])

num_df = pd.DataFrame(num_arr, columns=["A", "B", "C" ])
print(num_df)
```
