```python
# Cleaning ratings as rn it has integers, string and float values
def clean_data(data):
    # 1: Creating a map for ratings
    text_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
    cleaned_data = []
    unique_val = set() # for storing unique values

    for user in data:
        raw_rating = str(user["rating"]).strip().lower() # for removing any leading or trailing whitespace
        if(raw_rating in text_to_num):
            raw_rating = text_to_num[raw_rating]
        # print(raw_rating)
        user["rating"] = raw_rating

        # 2: Handling missing values
        # For age
        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None

        # 3: For duplicate values
        # can use deduplication technique here
        if(user["name"].strip() in unique_val):
            continue
        unique_val.add(user["name"].strip())
        cleaned_data.append(user)

    return cleaned_data
```

```python
data = clean_data(data)
print(data)
```

```python
# 2: Getting meaningful insights
# Finding avg rating
def avg_rating(data):
    total_rat = 0
    for user in data:
        total_rat += float(user["rating"])
    
    print(f"Total avg rating is {total_rat/len(data)}")

# % of users who gave poor rating(<3)
    poor_rat = 0
    for user in data:
        if(float(user["rating"]) < 3):
            poor_rat += 1
        
    print(f"Percentage of users who gave < 3 rating, {poor_rat/len(data)*100}%")
   
```

```python
avg_rating(data)
```

```python
# Building recommendation system
def recommend(data):
    recommendations = []

    for user in data:
        curr_recom = {}
        curr_recom["name"] = user["name"]
        if(float(user["rating"]) >=4):
            curr_recom["brand"] = "Apple"
        else:
            curr_recom["brand"] = "Samsung"
        recommendations.append(curr_recom)
    return recommendations
```

```python
recommend(data)
```
