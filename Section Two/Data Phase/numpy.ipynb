```python
import numpy as np
import time
```

```python
arr = np.array([1, 2, 3, 4, 5])
print(arr)
print(type(arr))
```

```python
# Checking Python list vs NumPy array in execution time
size = 100000000
li = list(range(size))
start = time.time() # It gives the current time in seconds
square = [x**2 for x in li]
end = time.time() 
print(f"Using Python list: {end - start} seconds")

# For numPy array
array = np.array(li)
start_time = time.time()
sqr = array ** 2
end_time = time.time()
print(f"Using NumPy array: {end_time - start_time} seconds")
```

```python
# memory usage in terms of bytes 
import sys
print(f"Memory usage of python list: {sys.getsizeof(li)} bytes")
# print(f"Memory usage of numPy array: {sys.getsizeof(array)} bytes")
# or
print(f"Memory usage of numPy array: {array.nbytes} bytes")
```
