## Scraping Activity

### Quotes Data Scraping

```python
import requests
from bs4 import BeautifulSoup
```

```python
# url = "https://quotes.toscrape.com/page/1/"
page_count = 1 # would show response 200 but actually the page doesn't exist

while True:
    url = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(url) # Now let's see how long the requests are going in

    soup = BeautifulSoup(res.text, "lxml")
    quotes = soup.select("div.quote")
    # print(quotes)
    # print(res)

    if not quotes:
        print("No valid pages anymore")
        break

    with open(f"Scrapped Data/quotes{page_count}.html", "w") as f:
        f.write(res.text)
        print(f"Downloaded pages from page {page_count}")

    page_count = page_count + 1
# Checking for valid-non-valid data
```
