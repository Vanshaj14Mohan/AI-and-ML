# Naruto Public API

```python
import requests
import pandas as pd
```

```python
URL = "https://dattebayo-api.onrender.com/characters"

req = requests.get(URL)
```

```python
req
```

```python
json_data = req.json() # Converting into json format
json_data
```

```python
df = pd.json_normalize(json_data["characters"]) # Making a dataFrame based on Characters
df.head()
```

```python
df = df[["name", "jutsu", "natureType", "tools"]] # Displaying selected columns
df.head()
```
