# Melt and Pivot part

```python
country_df = pd.DataFrame({
    "country": ["USA", "USA", "India", "India"],
    "year": [2020, 2021, 2020, 2021],
    "sales": [100, 120, 90, 110],
    "profit": [20, 25, 18, 22], 
    "tax": [2, 4, 5, 7]
})
```

```python
country_df
#Converting into long format by also preserving the values
```

```python
new_data = country_df.melt(
    id_vars= ["country", "year"],
    value_vars=["sales", "profit", "tax"],
    var_name = "metrices",
    value_name= "value"
)
```

```python
new_data
```

```python
# Using pivot to transfer to almost original data
original_one = new_data.pivot(
    index=["country", "year"],
    columns="metrices",
    values="value"
)
```

# Merging and Join Data

```python
df_customer = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["Adam", "Thomas", "John", "Jack"]
})

df_order = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": [2, 1, 4, 9],
    "amount": [100, 120, 90, 110]
})

# Data divided into 2 categories with customer_id in both dataFrame
```

```python
df_customer
```

```python
df_order
```

```python
pd.merge(df_customer, df_order, on="customer_id") #inner join
```

```python
pd.merge(df_customer, df_order, on="customer_id", how="left") #left join
```

```python
pd.merge(df_customer, df_order, on="customer_id", how="right") #right join
```

```python
pd.merge(df_customer, df_order, on="customer_id", how="outer") #outer join
```

# Data Concatenation with Pandas

```python
df1 = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["Adam", "Thomas", "Jack"]
})

df2 = pd.DataFrame({
    "id": [4, 5, 6],
    "name": ["John", "Dexter", "Henry"]
})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1, df2]) # row wise concatenation
```

```python
pd.concat([df1, df2], ignore_index=True) # here index value will not be preserved
```

```python
pd.concat([df1, df2], axis=1) # column wise concatenation
```
