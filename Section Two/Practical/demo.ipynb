# Data Transformation

```python
# Creating a new column in which we can calculate tax 
# income > 50k => 5% tax and <=50k then 2% tax
df2
```

```python
df2["income"] = df2["income"].fillna(df2["income"].mean())
```

```python
df2["tax"] = df2["income"].apply(lambda x: "5%" if x >= 50000 else "2%")
df2
```

```python
# Using map
gender_map = {"Male": "M", "Female": "F", "Unknown": "U"}
df2["gender"] = df2["gender"].map(gender_map)
```

```python
df2
```

```python
#using assign 
df2 = df2.assign(new_income = df2["income"] * 1.1)
```

```python
df2
```

```python
# Replacing values
#df2["country"] = df2.replace("USA", "US")
```

```python
df2["country"] = df2["country"].replace("USA", "US")
```
