```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 1:
scores = np.random.normal(loc=70, scale=10, size=200)
```

```python
mean_score = np.mean(scores)
mean_score
```

```python
plt.figure(figsize=(8, 5))
plt.hist(scores, bins=15, edgecolor="black")
plt.title("Histogram of Scores")
plt.xlabel("Scores")
plt.ylabel("Frequency")
plt.show()
```

```python
# 2: 
penguins = sns.load_dataset("penguins")
penguins.head()
```

```python
sns.lineplot(data=penguins, x="flipper_length_mm", y="body_mass_g", hue="species", errorbar=None)
plt.grid()
plt.title("Task 2")
plt.show()
```

```python
# 3: 
tips = sns.load_dataset("tips")
tips.head()
```

```python
sns.boxplot(data=tips, x="total_bill", y="day", hue="sex")
plt.title("Task 3")
plt.show()

# Outcome
# Sunday has the highest median total bill.
```

```python
# 4: 
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="sex", style="smoker")
plt.title("Task 4")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.grid()
plt.show()

# Outcome
# Smokers and non-smokers exhibit similar tipping behavior overall.
```

```python
tips.groupby("smoker", observed=False)["tip"].mean() # Extra part
```

```python
tips["tip_pct"] = tips["tip"] / tips["total_bill"] * 100

tips.groupby("smoker", observed=False)["tip_pct"].mean()
```

```python
tips_pivot = tips.pivot_table(index="day", columns="time", values="tip", aggfunc="mean", observed=False)
```
