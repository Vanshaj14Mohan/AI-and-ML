# Multiple dataset barchart

```python
movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]

revenue = [1005, 170, 427, 133, 232]

movies2 = [
    "The Wrestler",
    "Avatar",
    "Inception",
    "Hugo",
    "Lincoln"
]

movies2_revenue = [378, 2788, 892, 185, 175]

x=np.arange(len(years))
width=0.5
plt.figure(figsize=(8,6))
plt.bar(x-width/2, revenue, color="green",label="Oscar Movies", width=width) # Shift to left side
plt.bar(x+width/2, movies2_revenue, color="purple", label="Non Oscar Movies", width=width) # Shift to right side
plt.xlabel("Years")
plt.xticks(x, years) # Set x-ticks back to years
plt.ylabel("Revenue ($M)")
plt.title("Oscar vs Non-Oscar Movies")
plt.legend()
plt.show()
```

# Scatter plots


```python
# Ex 1:
people = ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J" ]

age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

plt.scatter(age, bp, cmap="viridis", c=bp, alpha=1)
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
plt.colorbar(label="Blood Pressure values") # To show mapping of colors to blood pressure values
plt.title("Age vs Blood Pressure")
for i in range(len(people)):
    plt.annotate(people[i], xy=(age[i], bp[i]), xytext=(age[i], bp[i]))
plt.show()
```

```python
# Ex 2:
people = ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J" ]

age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

colors = ["green" if x <130 else "red" for x in bp]

plt.scatter(age, bp, color=colors, alpha=0.7, marker="x")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
plt.colorbar()
plt.title("Age vs Blood Pressure")
plt.show()
```

```python
# Multiple Plots

cities = ["City A", "City B", "City C", "City D", "City E"]

# Winter: Temperature (°C) vs Humidity (%)
winter_temp = [5, 2, 10, 0, 7]
winter_humidity = [80, 75, 65, 85, 70]

# Summer: Temperature (°C) vs Humidity (%)
summer_temp = [25, 30, 28, 35, 27]
summer_humidity = [60, 50, 55, 45, 65]

plt.scatter(winter_temp, winter_humidity, label="Winter", color="blue" )
plt.scatter(summer_temp, summer_humidity, label="Summer", color="orange")

plt.xlabel("Temperature in celsius")
plt.ylabel("Humidity in %")
plt.title("Summer Vs Winter")
plt.legend()
plt.grid()
plt.show()
```

# Pie Charts

```python
# Pie Charts
# Ex 1:

expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]

plt.pie(amounts, labels=expenses, autopct="%1.1f%%")

plt.show()
```

```python
# Ex 2:
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]

explodes = [0, 0, 0, 0.1, 0] # To explode R&D slice by 10%
plt.pie(amounts, labels=expenses, autopct="%1.1f%%", wedgeprops = {
    "edgecolor": "black",
    "linewidth": 1,
    "linestyle": "--",
    "alpha": 0.9,
    "fill": True
},
explode= explodes,
startangle=90,
shadow=True,
)

plt.title("Company Expenses Chart", loc="center")
plt.show()
```
