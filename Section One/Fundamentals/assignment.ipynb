```python
# 1: WAP that asks the user for their name and age then print it
name = str(input("Enter user name: "))
age = int(input("Enter user age: "))

print("Hello", name, "Your age is: ", age)
```

```python
# 2: Take two numbers from user and print their sum, difference, product and quotient
a = int(input("Enter number one:"))
b = int(input("Enter number two:"))
print("A is:", a, "B is:", b)
print("Sum is:", a+b)
print("Product is:", a*b)
print("Difference is:", a-b)
print("Quotient is:",a/b)
```

```python
# 3: Ask the user to input two integers and one float and convert them all to floats and print their average.
a = int(input("Enter a"))
b = int(input("Enter b"))
c = float(input("Enter c"))
print("A", a)
print("B", b)
print("C", c)

avg = float(a+b+c)/3

print("Average is: ", avg)
```

```python
# 4: The user enters a string containing a number ("45"). convert it to integer, float and string again print their values and types
str_one = "45"
str_int = int(str_one)
print(str_int, type(str_int))

float_str = float(str_one)
print(float_str, type(float_str))

str_again = str(str_int)
print(str_again, type(str_again))
```

```python
# 5: Wap to swap values of two numbers entered by the user
a = int(input("Enter first number"))
b = int(input("Enter second number"))

print("A", a)
print("B", b)
print("After Swapping")

c = a
a = b
b =c

print("A", a)
print("B", b)
```

```python
# 6: Take the radius(r) as user input and print it's area
pie = 3.14
radius = float(input("Enter radius:"))
print("Radius is: ", radius)
area = pie * radius * radius
print("Area is: ", area)
```

```python
# 7: Write the formula for simple Interest P*R*T/100
p = float(input("Enter principal amount: "))
r = float(input("Enter rate: "))
t = int(input("Enter Time: "))
print("Principal is: ", p)
print("Rate is: ", r)
print("Time is: ", t)

SI = (p*r*t)/100

print("Simple Interest is: ", SI)
```

```python
# 8: Celsius to Fahrenheit conversion : Formula Fahren_temp = (Celsius_temp * (9/5)) + 32
# Take Celsius temp as string input convert it into float
celsius_temp = float(input("Enter temperature for celsius: "))
fahren_temp = (celsius_temp * 9/5) + 32
print("Celsius Temp", celsius_temp)

print(f"{celsius_temp} in Fahrenheit is:", fahren_temp)
```

```python
# 9: Take a decimal number as input like(45.78) and output it's integer and float point
#Integer part = > 45, Float Part => .78
num = float(input("Enter float number: "))
print("Number is: ", num)

int_num = int(num)

fac_num = num - int_num

print("Integer Part: ", int_num)
print("Factorial Part:", fac_num)
```
