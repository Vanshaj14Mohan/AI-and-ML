```python
# 1: WAP that asks the user for their name and age then print it
name = str(input("Enter user name: "))
age = int(input("Enter user age: "))

print("Hello", name, "Your age is: ", age)
```

```python
# 2: Take two numbers from user and print their sum, difference, product and quotient
a = int(input("Enter number one:"))
b = int(input("Enter number two:"))
print("A is:", a, "B is:", b)
print("Sum is:", a+b)
print("Product is:", a*b)
print("Difference is:", a-b)
print("Quotient is:",a/b)
```

```python
# 3: Ask the user to input two integers and one float and convert them all to floats and print their average.
a = int(input("Enter a"))
b = int(input("Enter b"))
c = float(input("Enter c"))
print("A", a)
print("B", b)
print("C", c)

avg = float(a+b+c)/3

print("Average is: ", avg)
```

```python
# 4: The user enters a string containing a number ("45"). convert it to integer, float and string again print their values and types
str_one = "45"
str_int = int(str_one)
print(str_int, type(str_int))

float_str = float(str_one)
print(float_str, type(float_str))

str_again = str(str_int)
print(str_again, type(str_again))


```
