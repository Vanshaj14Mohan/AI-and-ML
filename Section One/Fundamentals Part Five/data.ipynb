# JSON MODULE

```python
import json
json_str = '{"name": "John", "isStudent": true, "task": null}'
print(type(json_str))
# Converting into python objects
py_obj = json.loads(json_str)

print(py_obj)
print(type(py_obj))
```

```python
# Using json.dumps() for python object to json strings
data = {
    "name": "Jack",
    "isStudent": True,
    "task": None
}

json_str = json.dumps(data)
print(json_str)
print(type(json_str))
```
