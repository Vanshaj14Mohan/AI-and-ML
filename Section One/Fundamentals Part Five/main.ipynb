# File I/O in Python

```python
# For read method
f = open("sample.txt", "r") # returns a file object
data = f.read()
print(type(data)) # str
print(data)
f.close()
```

```python
# Now using write method on it
f = open("sample.txt", "w")
write = f.write("A normal sample file for Part Five based on File I/O in Python")
print(type(write))
f.close()
```

```python
# Using append mode
f = open("sample.txt", "a")
f.write("\nSome advanced features as well")
f.close()
```

```python
# Using x mode
f = open("sample2.txt", "x")
f.write("A new txt file to check more modes")
f.close()
```

```python
#using + mode with r,w and a
f = open("sample2.txt", "r+")
f.write("Python Language")
print(f.read())
f.close()
```

```python
# using a+
f = open("sample2.txt", "a+")
f.write("\nThe final part")
print(f.read()) # Nothing will be printed as file pointer is at the end
f.close()
```

```python
# using w+
f = open("sample2.txt", "w+")
f.write("The final part of the file")
print(f.read()) # Nothing will be printed as file pointer is at the end
f.close()
```

```python
# using with keyword
with open("sample2.txt", "r") as f:
    print(f.read())

# No need to explicitly close the file here, as it's handled by with keyword
```

```python
# A Word seacrh problem, Find word Python and print it's line number
data = True
line = 1
word = "Python"
with open("word.txt", "r") as f:
    while data:
        data = f.readline()

        if("Python" in data):
            print(f"{word} found at line number {line}")
            break

        print(data)
        line +=1
```

```python
# Exception Handling
try:
    n = int(input("Enter a number: "))
    print("Number entered:", n)
    ans = 10/n

except ZeroDivisionError:
    print(f"Divide by 0 not allowed")

except ValueError:
    print(f"Invalid Input")

else:
    print(f"ans is {ans}")

finally: # to execute all statements even if there is an error 
    print("End of the program")
    
# We can search for Built-in exceptions in Python and explore them.
```

```python
# List Comprehensions
# Print squares from 0-5 in a list
squares = []
for i in range(6):
    squares.append(i*i)

print(squares)

print("Using List Compreshensions") 
sq = [i*i for i in range(6)]
print(sq)

print("For even squares")
even = [i*i for i in range(6) if i%2==0]
print(even)

print("For odd case")
odd = [i*i for i in range(6) if i%2!=0]
print(odd)
```

```python
#Replace all -ve values with 0 using list compreshensions
nums = [-2, -4, 3, 5, 2, -1]
nums = [0 if val < 0 else val for val in nums]
print(nums)
```

```python
# Convert all lowercase into uppercase 
words = ["python", "data", "science", "journey"]
words = [val.upper() for val in words]
print(words)
```
