# File I/O in Python

```python
# For read method
f = open("sample.txt", "r") # returns a file object
data = f.read()
print(type(data)) # str
print(data)
f.close()
```

```python
# Now using write method on it
f = open("sample.txt", "w")
write = f.write("A normal sample file for Part Five based on File I/O in Python")
print(type(write))
f.close()
```

```python
# Using append mode
f = open("sample.txt", "a")
f.write("\nSome advanced features as well")
f.close()
```

```python
# Using x mode
f = open("sample2.txt", "x")
f.write("A new txt file to check more modes")
f.close()
```
