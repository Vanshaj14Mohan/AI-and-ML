# File I/O in Python

```python
# For read method
f = open("sample.txt", "r") # returns a file object
data = f.read()
print(type(data)) # str
print(data)
f.close()
```

```python
# Now using write method on it
f = open("sample.txt", "w")
write = f.write("A normal sample file for Part Five based on File I/O in Python")
print(type(write))
f.close()
```

```python
# Using append mode
f = open("sample.txt", "a")
f.write("\nSome advanced features as well")
f.close()
```

```python
# Using x mode
f = open("sample2.txt", "x")
f.write("A new txt file to check more modes")
f.close()
```

```python
#using + mode with r,w and a
f = open("sample2.txt", "r+")
f.write("Python Language")
print(f.read())
f.close()
```

```python
# using a+
f = open("sample2.txt", "a+")
f.write("\nThe final part")
print(f.read()) # Nothing will be printed as file pointer is at the end
f.close()
```

```python
# using w+
f = open("sample2.txt", "w+")
f.write("The final part of the file")
print(f.read()) # Nothing will be printed as file pointer is at the end
f.close()
```
