# File I/O Questions

```python
#1: 
with open("names.txt", "w") as f:
    f.write("John\n")
    f.write("Adam\n")
    f.write("Jackie\n")
    f.write("Thomas\n")
    f.write("Tom\n")
```

```python
# Now reading it 
with open("names.txt", "r") as f:
    data = f.read()
    print(data)
```
