```python
# 4: Write a program to check whether two lists share no common elements or not
# using sets here can use intersection to find common elements
list1 = [1, 2, 3, 4, 5]
list2 = [6, 7, 8, 9, 10]
# For common elements
set1 = set(list1)
set2 = set(list2)
common_elem = set1.intersection(set2)
if len(common_elem) == 0:
    print("No common elements")
else:
    print("Common elements are", common_elem)
```

```python
# 5: Ask the user to input a string and then check whether that string is palindrome or not
num = input("Enter a string..")

print("Entered string", num)
if num == num[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
# 9: Given a list of words and create a dictionary that maps each word to it's length.
words = ["apple", "banana", "kiwi", "grapes", "cherry", "mango"]
word_length = {word: len(word) for word in words}
print(word_length)
```

```python
# 10: Create a dictionary where keys = student names, values are marks(integer)
# Write a menu based program to perform various operations:
# A. Add a student, B. Update marks, C. Search for a student, D. Display all students and marks, E. Exit
students = {}
while True:
    print("\nMenu:")
    print("A, Add a student")
    print("B, Update marks")
    print("C, Search for a student")
    print("D, Display all students and marks")
    print("E, Exit")

    choice = input("Enter your choice from A-E:").upper()

    if choice == "A":
        name = input("Enter student name: ")
        marks = int(input("Enter students marks: "))
        students[name] = marks
        print("Students added")

    elif choice == "B":
        name = input("Enter student name to update marks: ")
        if name in students:
            marks = int(input("Enter new marks"))
            students[name] = marks
            print("Marks updated")
        else:
            print("Student not found")

    elif choice == "C":
        name = input("Enter student name to search:")
        if name in students:
            print(f"{name} has marks {students[name]}")
        else:
            print("Student not found")

    elif choice == "D":
        if students: 
            print("Students and marks:")
            for name, marks in students.items():
                print(f"{name}: {marks}")
            else:
                print("No students found")

    elif choice == "E":
        print("Existing program")
        break

    else:
        print("Invalid choice")

```
