## Strings Introduction

```python
#NOTE: Strings are immutable in python, once you create a string you can't change the values in it. 
word = "python"
#Indexing 
print(word[0]) # p
print(word[1]) # y

print(len(word)) # 6

# Backward indexing
print(word[-1]) # n
print(word[-2]) # 0

print("Using for loop to iterate through our string")
for seq in word:
    print(seq)
```

```python
# Slicing in strings
word = "Python programming"
print(word[: len(word)])

print(word[0:6]) # Python
print(word[7:18]) # programming
print(word[7:]) # programming
print(word[:6]) # Python
# Reverse the string
print(word[::-1]) # gnimmargorp nohtyP

print(word[::2]) # select every second character from starting: 

# Reverse printing
print(word[::-2]) # select every second character in reverse order: gmrp ohtP

print(word[-11: -2]) # programmi,  Reverse printing using neagative indexing
```

```python
# Formatting a string using f-string and format method
# Using format method
a = 5
b = 10

sum = a+b

print("Sum is: {}".format(sum))
# or
print("Sum of {} and {} is {}".format(a, b, sum))

# Another eg
name = "John"
age = 25
print("His name is {} and his age is {}".format(name,age)) 

print("Working on {}".format("Advanced AI/ML"))

# Index based formatting
print("Sum of {0} and {1} is {2}".format(a, b, sum))
print("Sum of {1} and {0} is {2}".format(a, b, sum))

# Value based formatting in this we can create variable and assign value to it and use it in format method
print("Values of {a} and {b} are".format(a=10, b=20))

# Using F String
# Literal String Interpolation means we can embed expressions inside string literals, using a minimal syntax.
c = 10
d = 20
avg = 10+20 /2
print(f"Average of {c} and {d} is:", avg)
```

```python
# Lists in Python
# Lists are mutable, ordered collection of items, can have different data types.
marks = [99, 98, 94, 95, "abc", 100.45]
print(type(marks)) # <class 'list'>
# Indexing in lists
print(marks[2]) # 98
print(marks[-1]) # 100.45
print(marks[-2]) # abc

# Reverse indexing using negative indexing
print(marks[-4: -2]) # [94, 95]

# Slicing in lists
print(marks[2:5]) # [94, 95, 'abc']
print(marks[:4]) # [99, 98, 94, 95]
print(marks[2:]) # [94, 95, 'abc', 100.45]
print(marks[::-1]) # [100.45, 'abc', 95, 94, 98, 99]

# List methods
nums = [1, 2, 3]
nums.append(4)
print(nums)
nums.insert(1, 1.5) # 1.5 at index 1
print(nums)

nums.sort() # In ascending order
print(nums)
nums.sort(reverse=True) # In descending order
print(nums)

nums.remove(1.5) 
print(nums)

data = [1,2,3,4,5, 5]
data.reverse() # Reverse the list
print(data)

data.reverse() # Reverse the list again to original order
data.remove(5) # Remove first occurrence of 5
print(data)

# Iterarting through a list
for num in data:
    print(num)

demo = [1, 2, 3, 4, 5, 6, 7]
# find 5 and print it's index
for i in range(len(demo)):
    if demo[i] == 5:
        print("Index of 5 is: ", i)
    else:
        print("Not found")

# Another way
x = 5
idx = 0
for val in demo:
    if (val == x):
        print(idx)
        break
    idx += 1

if 5 in demo:
    print(demo.index(5))
else:
    print("Not found")
```

```python
# Tuples in Python
# Immutable, ordered collection of items, can have different data types
# Works similar to lists, only difference is we can't change the values in tuple once we create it.
num = (1, 2, 3, 4, 5)
print(len(num)) # 5
print(type(num)) # <class 'tuple'>

# Indexing
print(num[3]) # 4
print(num[-1]) # 5

for val in num:
    print(val) 

# Accesing a particular value and it's index
for i in range(len(num)):
    if num[i] == 5:
        print("Index of 5 is:", i)
        break
else:
    print("Not found")

# Slicing in tuples
print(num[1:4]) # (2, 3, 4)
print(num[:3]) # (1, 2, 3)
print(num[::-1]) # (5, 4, 3, 2, 1)

# Calculating sum of all values in tuple
sum = 0
for val in num:
    sum+= val
    print("Total sum is: ", sum)

# Some methods in tuples
data = (1, 2, 3, 3, 3, 4, 5)
print(data.count(3)) # 3
print(data.index(3)) # 2
```
