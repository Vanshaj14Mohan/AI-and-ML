## Classes and Objects

```python
# Creating a student class
class Student:
    subject = "Python"
    college = "XYZ"
    year = 2026

# Now every object of Student class will have same values for all the class variables
# Creating objects of student class
stu1 = Student()
stu2 = Student()
print(stu1) # Will print memory address of stu1 object
print(stu2) # Will print memory address of stu2 object
# Memory address is not fixed and can change everytime we run the code

# Acccesing values of class variables
print(stu1.subject, stu1.college, stu1.year) # Python XYZ 2026
print(stu2.subject, stu2.college, stu2.year) # Same as above
```
