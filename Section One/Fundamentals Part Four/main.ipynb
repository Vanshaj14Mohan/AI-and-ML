## Classes and Objects

```python
# Creating a student class
class Student:
    subject = "Python"
    college = "XYZ"
    year = 2026

# Now every object of Student class will have same values for all the class variables
# Creating objects of student class
stu1 = Student()
stu2 = Student()
print(stu1) # Will print memory address of stu1 object
print(stu2) # Will print memory address of stu2 object
# Memory address is not fixed and can change everytime we run the code

# Acccesing values of class variables
print(stu1.subject, stu1.college, stu1.year) # Python XYZ 2026
print(stu2.subject, stu2.college, stu2.year) # Same as above
```

```python
# _init_method() a special method in python it's a constructor which is called when an object is created
class Teachers:
    def __init__(self):
        print("Constructor")

teach1 = Teachers()

# Another example
class Professor:
    def __init__(self, name, subject):
        self.name = name
        self.subject = subject
    
    def get_name(self):
        return self.name

prof1 = Professor("Dr. Drake", "Data Science")
prof2 = Professor("Dr. Richard", "Data Analytics")
print(prof1.name, prof1.subject)
print(prof2.name, prof2.subject)

print(f"{prof1.name} teaches {prof1.subject}")
print(f"{prof2.name} teaches {prof2.subject}")

print(prof1.get_name())
print(prof2.get_name())
```

```python
# Attributes class and instance attributes
class Employee:
    company = "ABC" # Class attribute
    def __init__(self, name, salary):
        self.name = name # Instance attribute
        self.salary = salary # Instance attribute

emp1 = Employee("John", 50000)
emp2 = Employee("Nick", 45000)

print(emp1.name, emp1.salary, emp1.company) # John 50000 ABC
print(emp2.name, emp2.salary, emp2.company) # Nick 45000 ABC
```

```python
# Instance Method
class Car:
    car_type = "Auto"

    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def car_info(self):
        print(f"{self.brand} has model {self.model} and is an {self.car_type} car")

   # using class methods
    @classmethod
    def car_detail(cls):
        print(f"All cars are {cls.car_type} cars")

    # using static method
    @staticmethod
    def car_discount(price, discount):
        final_price = price - (price * discount / 100) 
        print(f"Final price after discount is: {final_price}")
    
car1 = Car("Toyota", "Camry")
car2 = Car("Honda", "Civic")

car1.car_info() # Toyota has model Camry and is an Auto car
car2.car_info() # Honda has model Civic and is an Auto car

# Car.car_detail()
# or
car1.car_detail()

# Static method 
car1.car_discount(400000, 15)
```

```python
# Encapsulation in OOP's
class Bank:
    def __init__(self, name, balance, id):
        self.name = name # Public attribute
        self._balance = balance # Protected attribute using _ & for private __
        self.__id = id # Private attribute

# Creating getters and setters
    def get_id(self): # getters
        return self.__id

acc1 = Bank("David", 10000, 150)
print(acc1.name, acc1._balance, acc1._Bank__id)        
```

```python
# Inheritance in OOP's
class Employee: # Parent class
    start_time = "8am"
    end_time = "4am"

    def change_time(self, new_end_time): # getters and setter not necessary but good practice to use them
         self.end_time = new_end_time

class Teacher(Employee): # Child Class
        def __init__(self, subject):
            self.subject = subject

class Admin_Staff(Employee): # Another child class
     def __init__(self, role):
          self.role = role

t1 = Teacher("Math")
t1.change_time("5am")

s1 = Admin_Staff("Manager")
print(s1.role, s1.start_time, s1.end_time)
print(t1.subject, t1.start_time, t1.end_time)
```
