```python
# Practice Problem 
# Design & create an online store for Products(name, price). Track total products being created
# And create static method to calculate discount on each product based on a % parameter passed to it.

class Product:
    total_product = 0

    def __init__(self, name, price):
        self.name = name
        self.price = price
        Product.total_product+= 1

    def prod_info(self): # Instance method
        print(f"Price of {self.name} is Rs. {self.price}")

    @classmethod
    def count_product(cls): # Class Method
        print(f"Total products created: {cls.total_product}")

    @staticmethod
    def discount_product(price, discount):
        final_price = price - (price*discount/100)
        print(f"Final price after discount is: Rs. {final_price}")

p1 = Product("Laptop", 50000)
p2 = Product("Mobile", 45000)
p1.prod_info()
p2.prod_info()

Product.count_product()

#Static Method
p1.discount_product(p1.price, 10)
p2.discount_product(p2.price, 5)
```

```python
# 1: Create a bank account with attributes account_number, owner_name and balance
# Add methods to deposit, withdraw and check balance
class BankAccount:
    def __init__(self, account_number, owner_name, balance=0):
        self.account_number = account_number
        self.owner_name = owner_name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited Rs. {amount}. New balance is Rs. {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Not enough balance")
        else:
            print(f"Withdrawed Rs. {amount}. New balance = {self.balance - amount}")
            self.balance -= amount

    def check_balance(self):
        print(f"Current balance = Rs.{self.balance}")

acc1 = BankAccount("123567", "Jack", 10000)
acc1.deposit(5000)
acc1.withdraw(3000)
acc1.check_balance()
```

```python
# 2 Create a class Book with attributes title, author, list of reviews
# add methods to add a new review, count total reviews and display all reviews
class Book:
    def __init__(self, title, author, reviews=[]):
        self.title = title
        self.auhor = author
        self.reviews = []

    def review(self, review):
        self.reviews.append(review)
        print(f"Review added: {review}")

    def count_review(self):
        print(f"Total reviews: {len(self.reviews)}")

    def display_review(self):
        print("All reviews:")
        for review in self.reviews:
            print(f"{review}")

book1 = Book("Pride and Prejudice", "Jane Austen")
book1.review("A timesless classic!")
book1.review("A must read for everyone")

book1.count_review()
book1.display_review()

```

```python
# 3: 
class Student:
    def __init__(self, name, roll_no, marks):
        self.set_name(name)
        self.set_roll_no(roll_no)
        self.set_marks(marks)

    # Getter methods
    def get_name(self):
        return self.__name

    def get_roll_no(self):
        return self.__roll_no

    def get_marks(self):
        return self.__marks

    # Setter methods with validation
    def set_name(self, name):
        if name.strip() == "":
            print("Name cannot be empty")
        else:
            self.__name = name

    def set_roll_no(self, roll_no):
        if 1 <= roll_no <= 100:
            self.__roll_no = roll_no
        else:
            print("Roll number must be between 1 and 100")

    def set_marks(self, marks):
        if marks < 0:
            print("Marks cannot be negative")
        else:
            self.__marks = marks


# Example
s1 = Student("John", 45, 90)
print(s1.get_name(), s1.get_roll_no(), s1.get_marks())
```

```python
# 4: 
import math

class Shape:
    def area(self):
        print("Area not defined")


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height


# Example
c = Circle(5)
r = Rectangle(4, 6)
t = Triangle(3, 8)

print(c.area())
print(r.area())
print(t.area())
        
```

```python
# 5: 
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model


class Car(Vehicle):
    def __init__(self, brand, model, seats):
        super().__init__(brand, model)
        self.seats = seats


class Bike(Vehicle):
    def __init__(self, brand, model, engine_cc):
        super().__init__(brand, model)
        self.engine_cc = engine_cc


# Example
car1 = Car("Toyota", "Innova", 7)
bike1 = Bike("Yamaha", "R15", 155)

print(car1.brand, car1.model, car1.seats)
print(bike1.brand, bike1.model, bike1.engine_cc)
```

```python
# 6: 
from abc import ABC, abstractmethod

class Employee(ABC):

    @abstractmethod
    def calculate_salary(self):
        pass


class Intern(Employee):
    def __init__(self, stipend):
        self.stipend = stipend

    def calculate_salary(self):
        return self.stipend


class FullTimeEmployee(Employee):
    def __init__(self, salary, bonus):
        self.salary = salary
        self.bonus = bonus

    def calculate_salary(self):
        return self.salary + self.bonus


class ContractEmployee(Employee):
    def __init__(self, hours, rate):
        self.hours = hours
        self.rate = rate

    def calculate_salary(self):
        return self.hours * self.rate


# Example
i = Intern(10000)
f = FullTimeEmployee(50000, 5000)
c = ContractEmployee(40, 500)

print(i.calculate_salary())
print(f.calculate_salary())
print(c.calculate_salary())
```

```python
# 9:
class Herbivore:
    def eat_plants(self):
        print("Eats plants")


class Carnivore:
    def eat_meat(self):
        print("Eats meat")


class Omnivore:
    def eat_both(self):
        print("Eats both plants and meat")


class Bear(Herbivore, Carnivore, Omnivore):
    def __init__(self, name):
        self.name = name

    def show(self):
        print(f"I am a bear named {self.name}")


# Example
b = Bear("Baloo")

b.show()
b.eat_plants()
b.eat_meat()
b.eat_both()
```
