```python
# Practice Problem 
# Design & create an online store for Products(name, price). Track total products being created
# And create static method to calculate discount on each product based on a % parameter passed to it.

class Product:
    total_product = 0

    def __init__(self, name, price):
        self.name = name
        self.price = price
        Product.total_product+= 1

    def prod_info(self): # Instance method
        print(f"Price of {self.name} is Rs. {self.price}")

    @classmethod
    def count_product(cls): # Class Method
        print(f"Total products created: {cls.total_product}")

    @staticmethod
    def discount_product(price, discount):
        final_price = price - (price*discount/100)
        print(f"Final price after discount is: Rs. {final_price}")

p1 = Product("Laptop", 50000)
p2 = Product("Mobile", 45000)
p1.prod_info()
p2.prod_info()

Product.count_product()

#Static Method
p1.discount_product(p1.price, 10)
p2.discount_product(p2.price, 5)
```

```python
# 1: Create a bank account with attributes account_number, owner_name and balance
# Add methods to deposit, withdraw and check balance
class BankAccount:
    def __init__(self, account_number, owner_name, balance=0):
        self.account_number = account_number
        self.owner_name = owner_name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited Rs. {amount}. New balance is Rs. {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Not enough balance")
        else:
            print(f"Withdrawed Rs. {amount}. New balance = {self.balance - amount}")
            self.balance -= amount

    def check_balance(self):
        print(f"Current balance = Rs.{self.balance}")

acc1 = BankAccount("123567", "Jack", 10000)
acc1.deposit(5000)
acc1.withdraw(3000)
acc1.check_balance()
```

```python
# 2 Create a class Book with attributes title, author, list of reviews
# add methods to add a new review, count total reviews and display all reviews
class Book:
    def __init__(self, title, author, reviews=[]):
        self.title = title
        self.auhor = author
        self.reviews = []

    def review(self, review):
        self.reviews.append(review)
        print(f"Review added: {review}")

    def count_review(self):
        print(f"Total reviews: {len(self.reviews)}")

    def display_review(self):
        print("All reviews:")
        for review in self.reviews:
            print(f"{review}")

book1 = Book("Pride and Prejudice", "Jane Austen")
book1.review("A timesless classic!")
book1.review("A must read for everyone")

book1.count_review()
book1.display_review()

```
