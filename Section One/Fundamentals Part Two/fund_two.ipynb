```python
# Print if a number is even or odd
n = int(input("Enter any number"))
if(n%2 == 0):
    print("An even number")
else:
    print("Not an even number")
```

```python
# Creating a username and password setup using nesting
username = input("Enter username: ")
password = input("Enter password: ")

if(username == "admin" and password == "admin@1234"):
    print("Welcome User")
else:
    if(username != "admin"):
        print("Wrong Username")
    else:
        print("Wrong password")
```

```python
# Using match case
color = input("Enter a color: ")

match color:
    case "Red":
        print("Stop")
    case "Yellow":
        print("Look")
    case "Green":
        print("Go")
    case _:
        print("Wrong color")
```
