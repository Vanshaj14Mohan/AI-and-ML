```python
# Creating a username and password setup using nesting
username = input("Enter username: ")
password = input("Enter password: ")

if(username == "admin" and password == "admin@1234"):
    print("Welcome User")
else:
    if(username != "admin"):
        print("Wrong Username")
    else:
        print("Wrong password")
```

```python
# Using match case: Alternative for if-elif-else
color = input("Enter a color: ")

match color:
    case "Red":
        print("Stop")
    case "Yellow":
        print("Look")
    case "Green":
        print("Go")
    case _: #default case
        print("Wrong color")
```

```python
# Using loops
i = 1

while(i<=5):
    print("While loop", i)
    i+= 1

print("After loop execution", i)
```

```python
# Print numbers from 1-5 using while loop 
i = int(input("Enter number 1"))

while(i<=5):
    print(i)
    i+= 1

print("Updated I after execution", i)

print("For reverse case:")

j = int(input("Enter number 5"))

while(j>=1):
    print(j)
    j-=1

print("After execution j", j)
```

```python
# Create a multiplication table for any number n
n = int(input("Enter any number"))

i = 1

while(i<=10):
    print(n *i)
    i+=1

# using for loop
print("Using for loop")
num = int(input("Enter the nuumber"))
for i in range(1,11):
    print(f"{num} * {i} = ", num*i)

```

```python
# Break & Continue
i = 1
while(i<=10):
    if(i%6 == 0):
        break
    print(i)
    i+=1

print("Loop executed after finding 6")

# Skipping multiple of three
num = 1
while(num<=10):
    if(num%3 == 0):
        num+=1
        continue
    print(num)
    num+=1

print("Loop done")
```

```python
# Print all odd numbers using break and continue
i = 1

while(i<=10):
    print(i)
    i+=2

# Using continue
print("Using Continue")
num = 1
while(num<=10):
    if(num%2 == 0):
        num+= 1
        continue
    print(num)
    num+=1

# More simple
print("More simpler")
n = 0
while(n<10):
    n+=1
    if(n%2 == 0):
        continue
    print(n)
```

```python
# Using for loop
string = "python"

for var in string:
    print(var)

print("Checking if a alphabet exits in the string or not")

if "h" in string:
    print("h is present in the string")
else:
    print("Not exist")
```

```python
#Ranges
for i in range(1, 6):
    print(i)

word = "artificial intelligence" # Count the number of times i appears here
count = 0
for char in word:
    if (char == "i"):
        count+=1

print("Number of times i appears is: ", count)

# Print vowels in the string
counter = 0

for charc in word:
    if charc in "aeiou":
        counter+=1

print("Number of vowels in the string is: ", counter)
```

```python
# Using range to print even and odd numbers 
print("For Odd numbers")
for i in range(1, 11, 2):
    print(i)

print("For Even numbers")
for i in range(2, 11, 2):
    print(i)
```

```python
# Print sum of first n natural numbers
n = int(input("Enter any positive number"))

sum = 0

for i in range(1, n+1):
    sum+= i

print("Sum of first n natural number is:", sum)
```
