```python
# 2: Write a function that takes two integers a, b and print all even numbers between a and b (inclusive)
def even_no(a, b):
    for i in range(a, b+1):
        if(i%2 == 0):
            print(i)

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print(f"Even number between {a} and {b} is: ") 
even_no(a,b)
```

```python
# 3: Write a function that takes 1-100 values and print all numbers that are divisible by 3 and 5
def div_3_5():
    for i in range(1, 101):
        if(i%3 == 0 and i%5 == 0):
            print(i)

print("All numbers divisible by 3 and 5 between 1-100 are:")
div_3_5()
```
