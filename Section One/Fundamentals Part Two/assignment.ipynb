```python
# 2: Write a function that takes two integers a, b and print all even numbers between a and b (inclusive)
def even_no(a, b):
    for i in range(a, b+1):
        if(i%2 == 0):
            print(i)

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print(f"Even number between {a} and {b} is: ") 
even_no(a,b)
```

```python
# 3: Write a function that takes 1-100 values and print all numbers that are divisible by 3 and 5
def div_3_5():
    for i in range(1, 101):
        if(i%3 == 0 and i%5 == 0):
            print(i)

print("All numbers divisible by 3 and 5 between 1-100 are:")
div_3_5()
```

```python
# 5: making a simple calulator with operations +, -, *, /, %
def calculator(a, b, op):
    if(op == "+"):
        return a+b
    elif(op == "-"):
        return a-b
    elif(op == "*"):
        return a*b
    elif(op == "/"):
        return a/b
    elif(op == "%"):
        return a%b
    else:
        return "Inavlid operator"
    
cal = calculator(10, 5, "*")
print(cal)
    

```

```python
# 7: Function to check if a number is prime or not
def is_prime(n):
    if (n<=1):
        return False
    for i in range(2, n):
        if (n%i == 0):
            return False
    return True

n = int(input("Enter a number: "))
if is_prime(n):
    print(f"{n} is prime number")
else:
    print(f"{n} is not prime number")
```

```python
# 8: Write a program to continously take input a number n and print if it's positive or negative until user enters "quit"
while True:
    n = input("Enter a number or quit to exist:")
    if n.lower() == "quit":
        break
    n = int(n)
    if(n>0):
        print(f"{n} is a positive number")
    elif(n<0):
        print(f"{n} is a negative number")
    else:
        print(f"{n} is zero")
```

```python
# 9: Write a program to print the sum of digits of a number
n = int(input("Enter a number: "))
sum = 0
for digit in str(n):
    sum+= int(digit)

print(f"Sum of digits {n} is: ", sum)
```

```python
# 10: Write a function to return the count of the digits of a number
def count(n):
    count = 0
    for digit in str(n):
        count += 1
    return count

n = int(input("Enter a number: "))
print(f"Count of digits of {n} is, ", count(n))
```
