## Linear Regression Scratch Implementation

```python
import numpy as np
```

```python
# Linear Regression

class LinearRegression:
    # Creating a construtor as some values we need to initialize
    # Initialization Phase
    def __init__(self, learning_rate=0.01, n_iter=1000):
        self.bias = None
        self.weights = None
        self.lr = learning_rate
        self.n_iter = n_iter

    def fit(self, X, y): # X_Train, y_Train
        m, n = X.shape # Returns number of sample and number of features [rowxcolumn]
        # Step 1
        self.bias = 0
        self.weights = np.zeros(n) # as we don't know the number of features, so using array of zeros with n features
        
        # Gradient Descent 
        for i in range(self.n_iter):
            # Step 2 Calculate y_pred
            y_pred = self.bias + np.dot(X, self.weights)

            # Step 3 Calculate Gradient 
            db = (1/m) * np.sum(y_pred - y) # Dervivative of bias
            dw = (1/m) * np.dot(X.T, (y_pred - y))

            # Step 4 Applying convergence tp update weights and bias
            self.bias -= self.lr * db
            self.weights -= self.lr * dw

    def predict(self, X):
        y_pred = self.bias + np.dot(X, self.weights)
        return y_pred 

        
```

```python
# Testing Phase 
# Input data
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)
print("Predicted values:", y_pred)

print("Bias:", model.bias)
print("Weights:", model.weights)
```

# Linear Regression using OLS

```python
class LinearRegressionOLS():
    def __init__(self):
        self.bias = None
        self.weights = None

    def fit(self, X, y):
        m, n = X.shape
        X_b = np.c_[np.ones((m,1)), X] # Adding bias term to input data
        
        # Normal equation
        # theta = np.dot(np.linalg.inv(np.dot(X_b.T, X_b)), np.dot(X_b.T, y)) Another way to use this equation using @ operator
        theta = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y

        self.bias = theta[0]
        self.weights = theta[1:]
    
    def predict(self, X):
            y_pred = self.bias + np.dot(X, self.weights)
            return y_pred 
```

```python
# Testing Phase 
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegressionOLS()
model.fit(X, y)

y_pred = model.predict(X)
print("Predicted values:", y_pred)

print("Bias:", model.bias)
print("Weights:", model.weights)
```
