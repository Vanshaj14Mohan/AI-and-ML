## Logistic Regression Scratch Implementation

```python
# Same thing we did back in Linear Regression going to complete this in 4 steps same as we did in Linear Regression part.
```

```python
import numpy as np

class LogisticRegression:
    def __init__(self, learning_rate=0.01, n_iter=1000):
        self.bias = None
        self.weights = None
        self.lr = learning_rate
        self.n_iter = n_iter

    # A private function to calculate sigmoid value
    def _sigmoid(self, z):
        return (1 / (1 + np.exp(-z)))

    def fit(self, X, y):
        m, n = X.shape
        self.bias = 0
        self.weights = np.zeros(n)

        # Step 2 Gradient Descent
        for i in range(self.n_iter):
            z = self.bias + np.dot(X, self.weights)
            y_pred = self._sigmoid(z)

            # Step 3 Calculate Gradient 
            db = (1/m) * (y_pred - y) # Dervivative of bias
            dw = (1/m) * np.dot(X.T, (y_pred - y))

            # Step 4 Applying convergence tp update weights and bias
            self.bias -= self.lr * db
            self.weights -= self.lr * dw

    # A function to predict just probablities of the classes
    def get_prob(self, X):
        z = self.bias + np.dot(X, self.weights)
        return self._sigmoid(z)

    def predict(self, X, threshold=0.5):
        probab = self.get_prob(X)
        y_pred_bool = probab >= threshold
        return y_pred_bool.astype(int)
```

```python
X = np.array([[1, 2], [2, 3], [3, 4], [4, 5]])
y = np.array([0, 0, 1, 1])

model = LogisticRegression()
model.fit(X, y)

y_pred = model.predict(X)
print(y_pred)

print(model.get_prob( X))
```
